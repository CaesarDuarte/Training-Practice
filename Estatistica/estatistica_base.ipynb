## 1. Imports

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
SEED = 42
np.random.seed(SEED)
```

## 2. Dados

```python
# Gere dados sintéticos ou carregue um arquivo
# Exemplo sintético:
grupo_a = np.random.normal(loc=50, scale=10, size=200)
grupo_b = np.random.normal(loc=53, scale=10, size=200)
df = pd.DataFrame({'valor': np.concatenate([grupo_a, grupo_b]),
                   'grupo': ['A']*200 + ['B']*200})
df.head()
```

## 3. Estatísticas Descritivas

```python
df.groupby('grupo')['valor'].agg(['mean','median','std','min','max']).round(3)
```

## 4. Visualização da Distribuição

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Histograma com KDE
for g, ax in zip(['A','B'], axes):
    sns.histplot(df[df['grupo']==g]['valor'], kde=True, ax=ax, label=f'Grupo {g}')
    ax.set_title(f'Distribuição — Grupo {g}')

plt.tight_layout()
plt.show()

# Q-Q plot
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for i, (g, ax) in enumerate(zip(['A','B'], axes)):
    sm.qqplot(df[df['grupo']==g]['valor'], line='s', ax=ax)
    ax.set_title(f'Q-Q Plot — Grupo {g}')
plt.tight_layout()
plt.show()
```

## 5. Teste de Hipótese

**H0:** As médias dos grupos A e B são iguais  
**H1:** As médias diferem  
**Significância:** α = 0.05

```python
# Normalidade (Shapiro-Wilk)
for g in ['A', 'B']:
    stat, p = stats.shapiro(df[df['grupo']==g]['valor'])
    print(f'Shapiro Grupo {g}: stat={stat:.4f}, p={p:.4f}')

print()

# Teste t (paramétrico) ou Mann-Whitney (não paramétrico)
t_stat, p_valor = stats.ttest_ind(grupo_a, grupo_b)
print(f'Teste t: t={t_stat:.4f}, p={p_valor:.4f}')

alpha = 0.05
conclusao = 'Rejeita H0' if p_valor < alpha else 'Não rejeita H0'
print(f'Conclusão (α={alpha}): {conclusao}')
```

## 6. Tamanho do Efeito

```python
# Cohen's d
d = (grupo_a.mean() - grupo_b.mean()) / np.sqrt((grupo_a.std()**2 + grupo_b.std()**2) / 2)
print(f"Cohen's d: {d:.4f}")
print('Interpretação: pequeno<0.2 | médio<0.5 | grande>=0.8')
```

## 7. Intervalo de Confiança

```python
ic = stats.t.interval(0.95, df=len(grupo_a)-1,
                      loc=grupo_a.mean(),
                      scale=stats.sem(grupo_a))
print(f'IC 95% Grupo A: [{ic[0]:.2f}, {ic[1]:.2f}]')
```
